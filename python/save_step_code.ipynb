### Extract Custom Step SAS Code and UI JSON

This code reads all `.step` files within a folder and extracts the SAS code and UI JSON. By default, output locations are:

**SAS Code**: `/steps_dir/src/sas/<step name>.sas`

**UI Code**:  `/steps_dir/src/ui/<step name>.json`

### Setup

```python
import json
from pathlib import Path

steps_dir       = Path("Put Directory Here")
sas_output_dir  = steps_dir / 'src' / 'sas'
json_output_dir = steps_dir / 'src' / 'ui'
```

### Main script

Helper function for writing files consistently. Ensures that new lines are handled properly compared to Path.write_text()

```python
def write_to_file(file_path, content):
    with open(file_path, "w", encoding="utf-8", newline="") as f:
        f.write(content)
```

Create directories if they do not exist and read full JSON from .step file. 

Extract the SAS code and UI JSON, then write them to output locations.

```python
sas_output_dir.mkdir(exist_ok=True)
json_output_dir.mkdir(exist_ok=True)

files = steps_dir.glob(f'*.step') 

for file_path in files:
    with open(file_path, encoding="utf-8") as step:
        data      = json.load(step)
        sas_code  = data["templates"]["SAS"]
        json_code = data["ui"]

        output_sas_file  = sas_output_dir  / f"{file_path.stem}.sas"
        output_json_file = json_output_dir / f"{file_path.stem}.json"

        write_to_file(output_sas_file, sas_code)
        write_to_file(output_json_file, json_code)
```
