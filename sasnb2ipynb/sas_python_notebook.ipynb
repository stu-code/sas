# Build a model in Python, Score in SAS, Query with SQL

sasviya.ml models are designed to be fast, accurate, and Pythonic. These models can be converted to ASTOREs to easily fit as part of an overall SAS code flow or registered to Model Manager. This demo will walk through:
- A simple example of running Python code on a Viya server with a notebook
- Graphics support
- Modeling
- Switching from Python to SAS
- Switching from SAS to SQL

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

from datetime import date
from pathlib import Path
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sasviya.ml.tree import GradientBoostingClassifier

home   = Path(os.getenv('HOME'))
data   = home / 'data'
models = home / 'models'

SAS.symput("home", str(home))
SAS.symput("data", str(data))
SAS.symput("models", str(models))
```

```python
df = pd.read_parquet(data / 'hmeq.parquet')
df
```

```python
plt.clf()
sns.heatmap(df.select_dtypes(include='number').corr())
SAS.pyplot(plt)
```

```python
X = df.drop(['BAD'], axis=1)
y = df['BAD']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

numeric_features     = X.select_dtypes(include='number').columns.to_list()
categorical_features = X.select_dtypes(exclude='number').columns.to_list()

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())     
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ])

hmeq_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', GradientBoostingClassifier(random_state=42))
])

hmeq_pipeline
```

# Fitting a SAS Viya model and saving it as an ASTORE

```python
model = GradientBoostingClassifier(random_state=42)
model.fit(X_train, y_train)

# Save model as an ASTORE
model.export(models / 'hmeq_gboost_model.astore', replace=True)

# Print model details
model.customize_results()
```

```python
libname pq parquet "&data";

proc astore;
    upload store  = "&models/hmeq_gboost_model.astore"
           rstore = hmeq_gboost_model
    ;

    score rstore = hmeq_gboost_model 
          data   = pq.hmeq
          out    = hmeq_scored
    ;
run;
```
